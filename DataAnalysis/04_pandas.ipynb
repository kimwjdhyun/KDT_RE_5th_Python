### Series
- 인덱스(라벨)가 붙은 1차원 배열
- 인덱스 이름을 직접 지정할 수 있음

```python
# 시리즈 생성
# 1. 리스트로 생성
data = [10, 20, 30, 40]
s = pd.Series(data)
print(s)

# 0    10
# 1    20
# 2    30
# 3    40
# dtype: int64
```

```python
# 2. 인덱스를 직접 지정
s = pd.Series(data, index=["a", "b", "c", "d"])
print(s)

# a    10
# b    20
# c    30
# d    40
# dtype: int64
```

```python
# 3. 딕셔너리로 생성 
data = {'서울' : 100, '부산' : 200, '대구' : 300, '제주' : 400}
s = pd.Series(data)
print(s)

# 서울    100
# 부산    200
# 대구    300
# 제주    400
# dtype: int64
```

```python
# 4. 하나의 값으로 시리즈를 생성
s = pd.Series(7, index=["가", "나", "다"])
print(s)

# 가    7
# 나    7
# 다    7
# dtype: int64
```

```python
# 시리즈의 주요 속성
data = [10, 20, 30, 40]
s = pd.Series(data, index=['a', 'b', 'c', 'd'], name='test data')
print("values", s.values) # 실제 데이터 배열    values [10 20 30 40]
print("index", s.index)   # 인덱스 객체         index Index(['a', 'b', 'c', 'd'], dtype='object')
print("dtype", s.dtype)   # 실제 데이터의 타입  dtype int64
print("size", s.size)     # 데이터의 개수       size 4
print("shape", s.shape)   # 배열의 형태         shape (4,)
print("name", s.name)     # 시리즈의 이름       name test data
```

```python
# 시리즈의 연산
s1 = pd.Series([10, 20, 30], index=["a", "b", "c"])
s2 = pd.Series([5, 6, 7], index=["c", "b", "d"])

result = s1 + s2
print(result)

# a     NaN
# b    26.0
# c    35.0
# d     NaN
# dtype: float64
```

```python
# 브로드 캐스팅
s = pd.Series([5, 10, 15], index=['x', 'y', 'z'])

print(s + 3)
# x     8
# y    13
# z    18
# dtype: int64
print(s * 2)
# x    10
# y    20
# z    30
# dtype: int64
```

```python
# 불리언 연산 및 인덱싱
s = pd.Series([50, 60, 70, 80, 90, 100])

mask = s > 70
print(mask)
# 0    False
# 1    False
# 2    False
# 3     True
# 4     True
# 5     True
# dtype: bool
print(s[mask])
# 3     80
# 4     90
# 5    100
# dtype: int64
```

### DataFrame
- 2차원 라벨 데이터 구조
- Series를 모아놓은 것
- 행과 열이 있고, 행과 열에 대해 라벨을 부여할 수 있다.

```python
# DataFrame 생성

# 1. 딕셔너리로 생성
data = {
    "이름" : ['kim', 'lee', 'sin'],
    "나이" : [15, 25, 33],
    "도시" : ['서울', '부산', '대구']
}

df = pd.DataFrame(data)
df


#   이름	나이	 도시
# 0	kim	    15	    서울
# 1	lee	    25	    부산
# 2	sin	    33	    대구
```

```python
# 2. 리스트의 리스트(2차원 리스트)로 생성
data = [[1, "a"], [2, "b"], [3, "c"]]
df = pd.DataFrame(data, columns=["번호", "코드"])
df

#      번호    코드
#   0	1	    a
#   1	2	    b
#   2	3	    c
```

```python
# 3. 딕셔너리 리스트로 생성
data = [
    {'이름' : 'Kim', '나이' : 28, '도시' : '제주'},
    {'이름' : 'Choi', '나이' : 27, '도시' : '울산'}
]
df = pd.DataFrame(data, index=["bf", "gf"])
df
```

```python
# 4. Series의 딕셔너리로 생성
s1 = pd.Series(["국어", "수학", "영어"], index=["a", "b", "c"])
s2 = pd.Series([100, 90, 85], index=["a", "b", "c"])

df = pd.DataFrame({"과목" : s1, "점수" : s2})
df


#       과목	점수
#   a	국어	100
#   b	수학	90
#   c	영어	85
```
